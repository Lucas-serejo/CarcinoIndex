# Exploratory feature-versus-LS research (not implemented)

Clinical LS in the current MVP is supplied by the specialist. SAM generates a prompted segmentation mask; it does not infer LS. No classifier is implemented or demonstrated here.

Future feature-versus-LS modelling requires an adequate clinically reviewed dataset and a defined process for specialist-validated/corrected masks. Current segmentation attempts are not validated or reference masks.

Relative image features do not resolve monocular physical scale. This notebook makes no claims about physical lesion size, depth, clinical validation, or prediction performance.

Training and evaluation design remain future research work after these data and mask-review prerequisites are established.
